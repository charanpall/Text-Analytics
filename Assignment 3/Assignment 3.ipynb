```python
df_news = pd.read_csv('news_dataset.csv')

column_name = 'text' 

df_news['preprocessed_text'] = df_news[column_name].apply(preprocess_text)

display(df_news.head())
```

```python
vectorizer = CountVectorizer(max_df=0.95, min_df=2, max_features=1000)

X = vectorizer.fit_transform(df_news['preprocessed_text'])
feature_names = vectorizer.get_feature_names_out()

print(f"Document-Term Matrix shape: {X.shape}")
```

```python
no_of_topics = 5 

lda_model = LatentDirichletAllocation(n_components=no_of_topics, random_state=42)
lda_model.fit(X)

print("LDA Model Training Complete!")
```
